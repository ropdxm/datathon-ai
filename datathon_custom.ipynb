Bundled completed comparisons contain a highest original-fold development accuracy of **62.5%**. This is a posthoc development maximum across the listed policies. The results cell shows each measured policy; it does not establish independent custom-camera accuracy.

**All 39 custom recordings are currently unlabeled by a verified human.** The supplied custom workbook repeats the original labels and does not match these new filenames. This notebook creates **unverified local model proposals**, then runs an explicit train-only weak-label sensitivity comparison. It cannot certify their correctness.

A separate label-free experiment uses all custom recordings to train an unsupervised scaler/PCA representation while keeping classifier supervision entirely on original labels. Its matched projection baseline is reported separately.

The iPhone and Samsung videos are simultaneous views of the same scenes, with unresolved pairing. They remain one conservative training session. Only the original **40 unique recordings**, on their exact historical held-out folds, supply evaluation targets. No independent custom-camera test or 87% accuracy guarantee is established.

## 6. Generate proposals and train the two declared policies
The local Qwen annotator requests fixed numeric codes and timestamps from silent frames. It sees neither supplied labels nor predictions from the original trained heads. Unknown proposals remain excluded. Proposed positive labels contain a timestamp and are marked for human review.

The driver compares original-only training against original plus custom proposals at weight **0.25**, separately for the fixed semantic linear policy and the small point-supervised linear/RBF policy. Custom footage stays in training during every outer and inner split. Calibration and thresholds use only held-out original inner-fold labels. Accuracy, balanced accuracy, recall, specificity, F1/F2, ranking metrics and point-localization diagnostics are reported.

This can take many minutes. Local annotation has a hard **2,700-second** budget; partial completion is marked explicitly. Aggregate progress prints about every 30 seconds. Annotation time is not a claim about deployment latency.

## 7. Train using custom footage without inventing labels
Both arms fit a scaler/PCA32 on up to 24 uniformly selected windows per training source. One uses original training sources only; the other also uses the 39 unlabeled custom views. Candidate classifiers and thresholds use original inner-fold labels exclusively. No held-out original features enter projection fitting, and no custom labels enter any loss. The four candidate heads share each fitted projection. This adds actual unlabeled-data training even when every annotation proposal is unresolved. It can reduce accuracy; read the measured comparison before choosing a model.


```python
# Train with all custom recordings WITHOUT needing their class labels.
# A fresh guarded subprocess fits scaler/PCA on training windows only; original
# labels alone supervise heads and threshold selection on the frozen folds.
GUARDED_ADAPTATION = """
import runpy, sys
from pathlib import Path
workspace = Path(sys.argv[1])
sys.path.insert(0, str(workspace))
from tools.benchmark_semantic_vlm import configure_offline_environment, blocked_network
configure_offline_environment()
sys.argv = ['unlabeled_experiment.py', 'compare',
    '--original-manifest', str(workspace / 'runs/clip_b32/manifest.json'),
    '--unlabeled-manifest', str(workspace / 'runs/custom_v1/custom_unlabeled_manifest.json'),
    '--original-features', str(workspace / 'runs/clip_b32/features'),
    '--unlabeled-features', str(workspace / 'runs/custom_v1/clip_features'),
    '--folds-from', str(workspace / 'reference/clip_baseline_cv.json'),
    '--output-dir', str(workspace / 'runs/custom_v1/unlabeled_adaptation'), '--fit-final']
with blocked_network() as network:
    runpy.run_module('unlabeled_experiment', run_name='__main__')
assert network['attempts'] == 0
print('UNLABELED_NETWORK_ATTEMPTS', network['attempts'], flush=True)
"""
run_command([sys.executable, '-u', '-c', GUARDED_ADAPTATION, WORKSPACE])
```

## 8. Optional private train-owned notes
This disabled branch requires a notes ZIP derived locally from the original supplied notes, uploaded privately and extracted to `NOTES_DIR`. Notes remain unverified privileged training supervision. Owner checks prevent reading held-out-source notes during a fit; inference uses video features alone. Raw notes are excluded from result exports and never sent to a hosted AI service.


```python
# Optional branch: private, UNVERIFIED original-source notes.
# First create the private notes ZIP locally with the repository tools, then
# upload/extract it into this private runtime. Set NOTES_DIR to the directory
# containing index.json/by_source/<SHA>.json (or by_source itself).
# Do not send notes to a hosted AI service.
RUN_PRIVILEGED_NOTES = False
NOTES_DIR = BASE / 'private_original_notes'
PRIVILEGED_REPORT = WORKSPACE / 'runs/custom_v1/privileged_notes/privileged_cv_report.json'
PRIVILEGED_RUNTIME = PRIVILEGED_REPORT.parent / 'optional_runtime.json'
if RUN_PRIVILEGED_NOTES:
    assert NOTES_DIR.is_dir(), 'Set NOTES_DIR to privately extracted owner-SHA JSON files.'
    import time
    started = time.perf_counter()
    PRIVILEGED_REPORT.unlink(missing_ok=True)  # Do not reuse a stale completed report after failure.
    optional_status = {'status': 'failed', 'completed': False, 'notes_verified': False}
    try:
        run_command([sys.executable, '-u', 'privileged_experiment.py', 'compare',
            '--original-manifest', WORKSPACE / 'runs/clip_b32/manifest.json',
            '--features', WORKSPACE / 'runs/clip_b32/features', '--notes-dir', NOTES_DIR,
            '--note-cache', WORKSPACE / 'runs/custom_v1/privileged_note_cache',
            '--folds-from', WORKSPACE / 'reference/clip_baseline_cv.json',
            '--output-dir', PRIVILEGED_REPORT.parent, '--device', 'cuda', '--fit-final'])
        optional_status.update(status='completed', completed=True, returncode=0)
    except subprocess.CalledProcessError as error:
        optional_status.update(error_type=type(error).__name__, returncode=error.returncode)
    finally:
        optional_status['elapsed_seconds'] = time.perf_counter() - started
        PRIVILEGED_RUNTIME.parent.mkdir(parents=True, exist_ok=True)
        PRIVILEGED_RUNTIME.write_text(json.dumps(optional_status, indent=2))
    print('PRIVILEGED_OPTIONAL_STATUS', json.dumps(optional_status))
```
